```python
import yfinance as yf
from yfinance.scrapers.history import PriceHistory

# Create ticker
ticker = yf.Ticker("AAPL")

# Build PriceHistory manually
ph = PriceHistory(
    data=ticker._data,
    ticker=ticker.ticker,
    tz=ticker.fast_info["timezone"]
)

# Full historical OHLCV
hist = ph.history(period="6mo", interval="1d")
print("=== History ===")
print(hist.head())


# Custom date range
hist_custom = ph.history(start="2025-01-01", end="2025-05-01")
print("\n=== Custom Range ===")
print(hist_custom.head())


# Include actions
hist_actions = ph.history(period="1y", actions=True)
print("\n=== History with Actions ===")
print(hist_actions.head())


# Splits
splits = ph.get_splits(period="max")
print("\n=== Splits ===")
print(splits.tail())


# Dividends
dividends = ph.get_dividends(period="max")
print("\n=== Dividends ===")
print(dividends.tail())


# Capital gains (mostly ETFs/funds)
capital_gains = ph.get_capital_gains(period="max")
print("\n=== Capital Gains ===")
print(capital_gains.tail())


# All actions
actions = ph.get_actions(period="max")
print("\n=== All Actions ===")
print(actions.tail())


# Metadata
meta = ph.get_history_metadata()
print("\n=== Metadata ===")
print(meta)


# Intraday pre/post market
intraday = ph.history(period="5d", interval="1h", prepost=True)
print("\n=== Intraday ===")
print(intraday.head())


# Raw non-adjusted prices
raw = ph.history(period="1mo", auto_adjust=False)
print("\n=== Raw Prices ===")
print(raw.head())
```
